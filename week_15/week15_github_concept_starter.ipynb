## Step 1: Fill In Your Concept Information

Edit the text inside the quotation marks.

Keep the language beginner-friendly. The goal is to explain one concept clearly, not to sound advanced.

````python
# Change these values for your own concept artifact.
# Keep the triple quotes so your text can use more than one line.

concept_title = "False Positives and False Negatives"

what_this_means = """
A false positive happens when normal activity is incorrectly flagged as risky.
A false negative happens when risky activity is incorrectly treated as normal.
"""

why_it_matters = """
Security tools and rules are not perfect. If a threshold is too strict, analysts
may waste time reviewing normal activity. If a threshold is too loose, risky
activity may be missed.
"""

simple_example = """
A login system might flag an account after 10 failed login attempts. That rule
could catch suspicious activity, but it could also flag a real student who forgot
a password during registration week.
"""

what_can_go_wrong = """
If people trust alerts without context, they may overreact to normal behavior.
If people ignore alerts because there are too many, they may miss real risk.
"""

interview_language = """
I learned that security analysis involves tradeoffs. A rule can create false
positives or false negatives, so analysts need evidence, context, and clear
limitations before making a recommendation.
"""

course_connections = [
    "security analytics",
    "risk management",
    "data poisoning and model evasion",
    "privacy and ethics"
]
````

## Step 2: Optional Code Example

If your concept does not need code, you can leave this section as a simple text example.

If you want to include code, keep it short and heavily commented.

````python
# This tiny example shows how a threshold can create a review decision.
# It is not a real security tool. It is only a classroom explanation.

login_events = [
    {"user": "student_a", "failed_logins": 2, "context": "normal typo"},
    {"user": "student_b", "failed_logins": 12, "context": "forgot password"},
    {"user": "student_c", "failed_logins": 25, "context": "possible attack"},
]

# A threshold is a cutoff. Here, 10 or more failed logins means review.
threshold = 10

for event in login_events:
    # This comparison creates a True or False value.
    needs_review = event["failed_logins"] >= threshold

    # Print a readable result for each event.
    print(event["user"], "needs review?", needs_review, "| context:", event["context"])
````

## Step 3: Create Markdown Text

This cell turns your answers into Markdown.

Markdown is the format GitHub uses for `README.md` files.

````python
# Join the course connection list into Markdown bullet points.
connections_markdown = "\n".join(f"- {item}" for item in course_connections)

# Build a Markdown code block line by line.
# This avoids complicated quote nesting and makes the generated text easier to inspect.
code_lines = [
    '```python',
    'login_events = [',
    '    {"user": "student_a", "failed_logins": 2, "context": "normal typo"},',
    '    {"user": "student_b", "failed_logins": 12, "context": "forgot password"},',
    '    {"user": "student_c", "failed_logins": 25, "context": "possible attack"},',
    ']',
    '',
    'threshold = 10',
    '',
    'for event in login_events:',
    '    needs_review = event["failed_logins"] >= threshold',
    '    print(event["user"], "needs review?", needs_review, "| context:", event["context"])',
    '```',
]
code_block = "\n".join(code_lines)

# Build the README in sections.
# Each section becomes part of the final Markdown file.
readme_sections = [
    f"# {concept_title}",
    "",
    "## What This Means",
    what_this_means.strip(),
    "",
    "## Why It Matters",
    why_it_matters.strip(),
    "",
    "## Simple Example",
    simple_example.strip(),
    "",
    "## Tiny Code Example",
    code_block,
    "",
    "## What Can Go Wrong",
    what_can_go_wrong.strip(),
    "",
    "## How To Talk About This In An Interview",
    interview_language.strip(),
    "",
    "## Connection To CST 2412",
    connections_markdown,
    "",
    "## Limitation",
    "This is a beginner classroom explanation. It is meant to show the concept clearly, not to replace professional security tools or incident analysis.",
]

readme_text = "\n".join(readme_sections)

print(readme_text)
````

## Step 4: Save The README File

Run this cell to save the Markdown file.

In Colab, you can find the file in the file browser on the left.

````python
from pathlib import Path

# Save the generated Markdown text to a file.
# You can upload this file to GitHub as README.md.
output_file = Path("README_concept_portfolio.md")
output_file.write_text(readme_text, encoding="utf-8")

print("Saved:", output_file)
````
