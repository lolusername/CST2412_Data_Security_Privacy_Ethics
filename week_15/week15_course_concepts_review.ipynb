## Step 2: Print Concept Cards

The function below prints one concept in a readable format.

A function is a reusable block of code. Here, the function saves us from writing the same print statements over and over.

````python
def print_concept_card(item):
    """Print one course concept in a readable card format."""

    # The dictionary keys let us pull out each part of the concept record.
    print("=" * 72)
    print(item["concept"].upper())
    print("=" * 72)
    print("Plain meaning:")
    print(fill(item["plain_meaning"], width=72))
    print()
    print("Workplace phrase:")
    print(fill(item["workplace_phrase"], width=72))
    print()
    print("Interview sentence:")
    print(fill(item["interview_sentence"], width=72))
    print()
    print("Portfolio artifact idea:")
    print(fill(item["portfolio_artifact"], width=72))


# Print the first concept card as an example.
print_concept_card(course_concepts[0])
````

## Step 3: Search For Concepts

This helps you find a concept connected to a keyword like `privacy`, `cloud`, `logs`, `data`, or `access`.

````python
def search_concepts(keyword):
    """Return concept records that contain the keyword in any text field."""

    # Make the keyword lowercase so the search is not case-sensitive.
    keyword = keyword.lower()

    matches = []

    # Loop through each concept dictionary.
    for item in course_concepts:
        # Join all values in the dictionary into one searchable text string.
        combined_text = " ".join(item.values()).lower()

        # If the keyword appears anywhere in the text, keep this item.
        if keyword in combined_text:
            matches.append(item)

    return matches


# Change this keyword to search for another topic.
keyword = "privacy"
results = search_concepts(keyword)

print(f"Search keyword: {keyword}")
print(f"Matches found: {len(results)}")
print()

for item in results:
    print("-", item["concept"])
````

## Step 4: Choose One Concept For Your GitHub Artifact

Change `chosen_concept_name` to one concept from the list.

Then run the cell to print a concept card you can use as a starting point.

````python
# Change this value to another concept if you want.
chosen_concept_name = "Privacy incident response"

# Start with no match.
chosen = None

# Loop through the course concepts until we find the selected concept.
for item in course_concepts:
    if item["concept"] == chosen_concept_name:
        chosen = item
        break

# If the concept was found, print it. If not, print a helpful message.
if chosen is None:
    print("Concept not found. Check the spelling or choose a concept from the list above.")
else:
    print_concept_card(chosen)
````

## Step 5: Generate Resume Bullet Starters

These are not meant to be copied blindly.

Use them as starting points, then make them honest and specific to your actual work.

````python
def make_resume_bullet(item):
    """Create a resume-style bullet starter from a concept record."""

    # This is a template string called an f-string.
    # It lets us insert dictionary values into a sentence.
    return (
        f"Explained {item['concept'].lower()} using beginner-friendly examples, "
        f"with emphasis on {item['workplace_phrase']}."
    )


for item in course_concepts:
    print("-", make_resume_bullet(item))
````

## Step 6: Export A Small Portfolio Planning CSV

This exports a CSV with concept names, interview sentences, and portfolio ideas.

You can upload the CSV to GitHub, use it while planning your final lab, or copy one row into your README.

````python
# We choose which dictionary keys should become CSV columns.
fieldnames = ["concept", "interview_sentence", "portfolio_artifact"]

# Open the CSV file for writing.
# newline="" prevents blank lines between rows on some systems.
with output_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    # Write the header row first.
    writer.writeheader()

    # Write one row for each concept.
    for item in course_concepts:
        writer.writerow({
            "concept": item["concept"],
            "interview_sentence": item["interview_sentence"],
            "portfolio_artifact": item["portfolio_artifact"],
        })

print("Saved:", output_path)
````
